## Etape construction de l'arborescence

```python
# Création d'une variable qui va stocker le résultat du challenge
resultatChallenge = 0

class node():
    def __init__(self, nom, isDir, parent, size):
        self.nom = nom
        self.isDir = isDir
        self.size = size
        self.children = []
        self.parent = parent

    def calculerTaille(self):
        if self.isDir:
            self.size = sum([c.calculerTaille() for c in self.children])
        return self.size

    def __eq__(self, other):
        return self.nom == other.nom

    def __str__(self):
        if self.isDir:
            return f"dir : {self.nom} children : {len(self.children)}"
        else:
            return f"file : {self.nom} {self.size}"
racine = node("Racine", True, None , 0)
currentNode = None

listdir = [racine]

# Pour chaque instruction des lignes du fichier
for instruction in lignes:
    if verbose == True: print(f"{instruction=}")

    if instruction.startswith("$ cd /") :
        if verbose == True: print(f"  > On remonte à la racine")
        currentNode = racine
        continue

    if instruction.startswith("$ cd ..") :
        if verbose == True: print(f"  > On remonte au parent")
        currentNode = currentNode.parent
        continue

    if instruction.startswith("$ ls") :
        continue

    if instruction.startswith("$ cd"):
        for child in currentNode.children:
            if child.nom == instruction.split(" ")[2].strip():
                currentNode = child
                break
        if verbose == True: print(f"  > On descend vers {currentNode.nom}")
        continue

    if instruction.startswith("dir") :
        nameDir = instruction.split(" ")[1].strip()
        childNode = node(nameDir, True, currentNode , 0)
        listdir.append(childNode)
        currentNode.children.append(childNode)
        if verbose == True: print(f"  > Ajout du répertoire enfant {nameDir}")
        continue

    decoup = instruction.split(" ")
    currentNode.children.append(node(decoup[1].strip(), False, currentNode , int(decoup[0])))
    if verbose: print(f"  > Ajout du fichier enfant {decoup[1].strip()} - taille : {decoup[0]}")

racine.calculerTaille()

print(f"\nRésultat du challenge partie 1 : {resultatChallenge}")
```

## Etape 1

On a construit l'arborescence ainsi que la liste des dossiers.
On va récuperer la liste des dossiers dont la taille sont inférieur à 100000 et en faire la somme

```python

resultatChallenge = sum([c.size for c in listdir if c.size <= 100000])

print(f"\nRésultat du challenge partie 1 : {resultatChallenge}")
```

## Etape 2 Recherche du nombre de fois ou on passe par 0

```python
# Création d'une variable qui va stocker le résultat du challenge
resultatChallenge = 0

totalSpace = 70000000

dispo = totalSpace-racine.size
print(f"Espace disponible : {dispo}")

tailleUpdate = 30000000

tailleASup = tailleUpdate - dispo

print(f"Taille à supprimer : {tailleASup}")

listdir = sorted(listdir, key=lambda x: x.size)

for dir in listdir:
    print(f" {dir.nom}: {dir.size}")
    if dir.size > tailleASup:
        resultatChallenge = dir.size
        break


print(f"\nRésultat du challenge partie 2 : {resultatChallenge}")
```
