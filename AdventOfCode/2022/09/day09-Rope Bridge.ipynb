## Utilisation des données du challenges

```python
fichier = open("input_challenge.txt", "r")
lignes = fichier.readlines()

# Convertir chaque nombre en une liste de chiffres
data = [["." for i in range(1000)] for k in range(1000)]
posXHead = 500
posYHead = 500
posXTail = 500
posYTail = 500

verbose = False
```

---
## Etape 1

```python
# Création d'une variable qui va stocker le résultat du challenge
resultatChallenge = 0

import pandas as pd


# Créer un DataFrame
df = pd.DataFrame(data)
df2= pd.DataFrame(data)



df.iloc[posXHead, posYHead] = "H"


df2.iloc[posXTail, posYTail] = "x"
# print(df2)

# Pour chaque instruction des lignes du fichier
for instructions in lignes:
    direction,pas = instructions.split(" ")
    if verbose == True: print(f"{direction=} {pas=}")
    for i in range(int(pas)):
        df.iloc[posXHead, posYHead] = "."
        df.iloc[posXTail, posYTail] = "."


        match (direction):
            case "R":
                if abs(posYHead + 1 - posYTail) > 1:
                    posXTail = posXHead
                    posYTail = posYHead
                posYHead += 1
            case "L":
                if abs(posYHead -1 - posYTail) > 1:
                    posXTail = posXHead
                    posYTail = posYHead
                posYHead -= 1
            case "U":
                if abs(posXHead -1 - posXTail) > 1:
                    posXTail = posXHead
                    posYTail = posYHead
                posXHead -= 1
            case "D":
                if abs(posXHead -1 - posXTail) > 1:
                    posXTail = posXHead
                    posYTail = posYHead
                posXHead += 1

        df.iloc[posXHead, posYHead] = "H"
        df.iloc[posXTail, posYTail] = "T"
        df2.iloc[posXTail, posYTail] = "x"
        if verbose:
            print(df)
            print(df2)
            print("")
            print("")

sum = (df2 == "x").sum().sum()
print(f"\nRésultat du challenge partie 1 : {sum}")
```
