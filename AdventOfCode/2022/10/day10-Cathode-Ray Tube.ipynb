## Utilisation des données du challenges

```python
fichier = open("input_challenge.txt", "r")
lignes = fichier.readlines()

verbose = False
```

---
## Etape 1

```python
# Création d'une variable qui va stocker le résultat du challenge
resultatChallenge = 0

data = lignes.copy()


valueX = 1
signalStrength = 0

cptInstructions = 0

isWorking = False
addx = 0
# Pour chaque instruction des lignes du fichier
for i in range(1,221):

    if i in [20,60,100,140,180,220]:
        signalStrength = valueX * i
        if verbose: print(f" {i} : {signalStrength=}")
        resultatChallenge += signalStrength

    if not isWorking:
        instruction = data.pop(0)
        if verbose: print(f" {i=} : {instruction=}")
        if instruction.startswith("addx"):
            addx = int(instruction.split(" ")[1])
            isWorking = True
        # Si on rentre pas dans addx, on a une instruction noop et on a rien a faire
    else:
        valueX += addx
        if verbose: print(f" {i=}  > added {valueX=}")
        isWorking = False



print(f"\nRésultat du challenge partie 1 : {resultatChallenge}")
```

---
## Etape 2

```python
# Création d'une variable qui va stocker le résultat du challenge
resultatChallenge = 0

data = lignes.copy()

sprite = [0,1,2]
board = []

cptInstructions = 0

isWorking = False
addx = 0

ligne = ""
# Pour chaque instruction des lignes du fichier
for i in range(0,240):

    if verbose: print(f" {i=} {(i % 40)=} {sprite=}")

    if ((i) % 40) in sprite:
       ligne += "#"
    else:
        ligne += "."

    if len(ligne) == 40:
        board.append(ligne)
        ligne = ""

    if not isWorking:
        instruction = data.pop(0)
        if verbose: print(f" {i=} : {instruction=}")
        if instruction.startswith("addx"):
            addx = int(instruction.split(" ")[1])
            isWorking = True
        # Si on rentre pas dans addx, on a une instruction noop et on a rien a faire
    else:
        sprite = [value + addx for value in sprite]
        if verbose: print(f" {i=}  > added {sprite=}")
        isWorking = False

print("")
for v in board:
    print(v)

print(f"\nRésultat du challenge partie 1 : {resultatChallenge}")
```
