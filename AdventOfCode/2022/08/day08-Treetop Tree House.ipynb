## Chargement avec panda

```python
import pandas as pd

# Convertir chaque nombre en une liste de chiffres
data = [[int(digit) for digit in ligne.strip()] for ligne in lignes]

# Créer un DataFrame
df = pd.DataFrame(data)

print(df)
```

---
## Etape 1

```python
# Création d'une variable qui va stocker le résultat du challenge
resultatChallenge = 0

# On commence par ajouter les arbres qui font le tour
resultatChallenge += df.shape[0] * 2
resultatChallenge += df.shape[1] * 2 - 4

# On va tester les arbres intérieurs
for i in range(1,df.shape[0]-1):  # Parcourir les lignes
    for j in range(1,df.shape[1]-1):  # Parcourir les colonnes
        # Tester un arbre
        if verbose: print(f"  Test arbre  {i}:{j} - taille {df.iloc[i, j]}")

        # au gauche
        isVisible = True
        for k in range(0,j):
            if verbose: print(f"    > gauche {i}:{k} - taille {df.iloc[i, k]}")
            if df.iloc[i, k] >= df.iloc[i, j]:
                if verbose: print(f"    > gauche {i}:{k} - cache l'arbre")
                isVisible = False
                break

        if isVisible:
            if verbose: print(f"    > Arbre visible par la gauche")
            resultatChallenge += 1
            continue

        # En droite
        isVisible = True

        for k in range(j+1,df.shape[1]):
            if verbose: print(f"    > droite {i}:{k} - taille {df.iloc[i, k]}")
            if df.iloc[i, k] >= df.iloc[i, j]:
                if verbose: print(f"    > gauche {i}:{k} - cache l'arbre")
                isVisible = False
                break

        if isVisible:
            if verbose: print(f"    > Arbre visible par la droite")
            resultatChallenge += 1
            continue

        # A dessus
        isVisible = True

        for k in range(0,i):
            if verbose: print(f"    > dessus {k}:{j} - taille {df.iloc[k, j]}")
            if df.iloc[k, j] >= df.iloc[i, j]:
                if verbose: print(f"    > gauche {k}:{j} - cache l'arbre")
                isVisible = False
                break

        if isVisible:
            if verbose: print(f"    > Arbre visible par la dessus")
            resultatChallenge += 1
            continue

        # A dessous
        isVisible = True

        for k in range(i+1,df.shape[0]):
            if verbose: print(f"    > dessous {k}:{j} - taille {df.iloc[k, j]}")
            if df.iloc[k, j] >= df.iloc[i, j]:
                if verbose: print(f"    > gauche {k}:{j} - cache l'arbre")
                isVisible = False
                break

        if isVisible:
            if verbose: print(f"    > Arbre visible par la dessous")
            resultatChallenge += 1
            continue




print(f"\nRésultat du challenge partie 1 : {resultatChallenge}")
```

## Etape 2

```python
# Création d'une variable qui va stocker le résultat du challenge
resultatChallenge = 0

maxScenicScore = 0
# On va tester les arbres intérieurs
for i in range(1,df.shape[0]-1):  # Parcourir les lignes
    for j in range(1,df.shape[1]-1):  # Parcourir les colonnes
        # Tester un arbre
        if verbose: print(f"  Test arbre  {i}:{j} - taille {df.iloc[i, j]}")

        gauche = 0
        droite = 0
        dessus = 0
        dessous =0

        # au gauche
        for k in range(j-1,-1,-1):
            gauche += 1
            if df.iloc[i, k] < df.iloc[i, j]:
                if verbose: print(f"    > gauche {i}:{k} taille {df.iloc[i, k]} - plus petit -> visible")
            else:
                if verbose: print(f"    > gauche {i}:{k} taille {df.iloc[i, k]} - Egal ou sup-> visible + block")
                break

        # En droite
        for k in range(j+1,df.shape[1]):
            droite += 1
            if df.iloc[i, k] < df.iloc[i, j]:
                if verbose: print(f"    > droite {i}:{k} taille {df.iloc[i, k]} - plus petit -> visible")
            else:
                if verbose: print(f"    > droite {i}:{k} taille {df.iloc[i, k]} - Egal ou sup-> visible + block")
                break

        # A dessus

        for k in range(i-1,-1,-1):
            dessus += 1
            if df.iloc[k, j] < df.iloc[i, j]:
                if verbose: print(f"    > dessus {k}:{j} taille {df.iloc[k, j]} - plus petit -> visible")
            else:
                if verbose: print(f"    > dessus {k}:{j} taille {df.iloc[k, j]} - Egal ou sup-> visible + block")
                break

        # A dessous
        for k in range(i+1,df.shape[0]):
            dessous += 1
            if df.iloc[k, j] < df.iloc[i, j]:
                if verbose: print(f"    > dessous {k}:{j} taille {df.iloc[k, j]} - plus petit -> visible")
            else:
                if verbose: print(f"    > dessous {k}:{j} taille {df.iloc[k, j]} - Egal ou sup-> visible + block")
                break

        scenicScore = gauche * droite * dessus * dessous
        if verbose: print(f"    >> {scenicScore=}\n")

        if scenicScore > maxScenicScore:
            maxScenicScore = scenicScore


print(f"\nRésultat du challenge partie 2 : {maxScenicScore}")
```
