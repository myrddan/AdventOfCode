## Etape 2

Confident that your list of box IDs is complete, you're ready to find the boxes full of prototype fabric.

The boxes will have IDs which differ by exactly one character at the same position in both strings. For example, given the following box IDs:

    abcde
    fghij
    klmno
    pqrst
    fguij
    axcye
    wvxyz

The IDs abcde and axcye are close, but they differ by two characters (the second and fourth). However, the IDs fghij and fguij differ by exactly one character, the third (h and u). Those must be the correct boxes.

What letters are common between the two correct box IDs? (In the example above, this is found by removing the differing character from either ID, producing fgij.)

## Utilisation des données d'entrainements etape 2

```python
fichier = open("input_test2.txt", "r")
lignes = fichier.readlines()

verbose = True
```

```python
from itertools import combinations

# Création d'une variable qui va stocker le résultat du challenge
resultatChallenge = 0

# Pour chaque instruction des lignes du fichier
for a,b in combinations(lignes,2):
    # On crée une variable qui va  stock l'index de la différence
    # Losqu'on trouve une différence s'il la valeur est -1 alors c'est le premier qu'on trouve sinon c'est
    diff = -1
    for i in range(len(a)-1):
        if a[i] != b[i]:
            # A la première différence je maj diff
            if diff == -1:
                diff = i
            # Si je trouve une seconde différence, alors je remet diff à -1 pour qu'on n'execute pas
            # le code d'affichage du résultat et j'arrête de chercher sur ces deux id
            else:
                diff = -1
                break

    # Si la valeurs est différente de -1 c'est que j'ai trouvé une différence et une seul
    if diff != -1:
        # Une des façon de supprimer un caractère d'une chaine de caractère
        # est de le transfermer en liste, supprimer avec l'index puis retransformer en str
        l = list(a)
        del l[diff]
        resultatChallenge = "".join(l)
        
        print(f"\nRésultat du challenge partie 2 : {resultatChallenge}")
```
