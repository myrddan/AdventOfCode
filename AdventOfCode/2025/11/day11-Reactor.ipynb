En changeant juste le point de départ puis en filtrant, la solution semble simple et rapide mais l'algorythme se fige. Signe que la profondeur de la recherche surpasse les capacités de récursivités de python.

Je pense alors utiliser un DFS itératif. Nouvelle échec, la solution est d'utiliser la fonction @cache sur une DFS récursif en remplacant la liste des chemins valide par un compteur.

## Etape 1 : Recherche de YOU à OUT.

```python
# Création d'une variable qui va stocker le résultat du challenge
resultatChallenge = 0


# Initialiser un dictionnaire vide
datas = {}

# Parcourir chaque ligne du texte
for ligne in lignes:
    # Séparer la clé et les valeurs
    cle, *valeurs = ligne.split(':')
    # Nettoyer les espaces et ajouter au dictionnaire
    datas[cle.strip()] = [v.strip() for v in valeurs[0].split()]

if verbose:print(datas)

def trouver_chemins_DFS_recursif(depart, arrivee, chemin_en_cours=[], chemins_valides=[]):
    chemin_en_cours = chemin_en_cours + [depart]
    if depart == arrivee:
        chemins_valides.append(chemin_en_cours)
    else:
        for pas_suivant in datas[depart]:
            if pas_suivant not in chemin_en_cours:  # Évite les cycles
                chemins_valides = trouver_chemins_DFS_recursif(pas_suivant, arrivee, chemin_en_cours, chemins_valides)

    return chemins_valides


result = trouver_chemins_DFS_recursif('you', 'out',[],[])

if verbose:
    for chemin in result:
        print(" → ".join(chemin))

resultatChallenge = len(result)

print(f"\nRésultat du challenge partie 1 : {resultatChallenge}")
```

```python
# Création d'une variable qui va stocker le résultat du challenge
resultatChallenge = 0


# Initialiser un dictionnaire vide
datas = {}

# Parcourir chaque ligne du texte
for ligne in lignes:
    # Séparer la clé et les valeurs
    cle, *valeurs = ligne.split(':')
    # Nettoyer les espaces et ajouter au dictionnaire
    datas[cle.strip()] = [v.strip() for v in valeurs[0].split()]


if verbose:print(datas)


def trouver_chemins_DFS_iteratif(graphe, depart, arrivee):

    chemins_valides = []
    # On va lors du parcours ajouter les noeuds dans une pile LIFO
    pile = [(depart, [depart])]  # (noeud_courant, chemin_actuel)

    # Tant que la pile n'est pas vide alors on continue la recherche
    while pile:
        # On prend la dernière valeur ajouté
        noeud, chemin = pile.pop()
        if noeud == arrivee:
            if "fft" in chemin and "dac" in chemin:
                if verbose: print(" → ".join(chemin))
                chemins_valides.append(chemin)
        else:
            for voisin in graphe[noeud]:
                if voisin not in chemin:  # Évite les cycles
                    pile.append((voisin, chemin + [voisin]))

    return chemins_valides


result = trouver_chemins_DFS_iteratif(datas, 'svr', 'out')


resultatChallenge = len(result)

print(f"\nRésultat du challenge partie 1 : {resultatChallenge}")
```
